```python
steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]

mean = sum(steps) / len(steps)

squared_deviations = [(x - mean) ** 2 for x in steps]

sum_squared_deviations = sum(squared_deviations)

variance = sum_squared_deviations / len(steps)

print("Mean:", mean)
print("Variance:", variance)

```

```python


def calculate_standard_deviation(scores):
    return round(statistics.pstdev(scores), 2)

scores = [80, 85, 90, 75, 95]

print(calculate_standard_deviation(scores))
```

```python


friend_A = [200, 200, 200, 200, 200]
friend_B = [100, 300, 150, 400, 50]

std_A = statistics.pstdev(friend_A)
std_B = statistics.pstdev(friend_B)

print("Friend A Standard Deviation:", std_A)
print("Friend B Standard Deviation:", round(std_B, 2))
```
