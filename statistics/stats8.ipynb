```python
def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes
probability_heads = calculate_probability(1, 2)

print("Probability of getting Heads:", probability_heads)
```

```python
import random

count_six = 0

for i in range(100):
    roll = random.randint(1, 6)
    
    if roll == 6:
        count_six += 1

probability = count_six / 100

print("Number of times 6 appeared:", count_six)
print("Experimental probability of getting 6:", probability)
```

```python
genres = ['pop', 'rock', 'pop', 'hiphop', 'pop',
          'jazz', 'rock', 'pop', 'hiphop', 'pop',
          'rock', 'pop', 'jazz', 'pop', 'rock',
          'hiphop', 'pop', 'rock', 'pop', 'jazz']

pop_count = genres.count('pop')
total_songs = len(genres)

probability = pop_count / total_songs

print("Number of pop songs:", pop_count)
print("Probability of picking a pop song:", probability)
```

```python
def calculate_probability(food_users, dessert_users):
    food_set = set(food_users)
    dessert_set = set(dessert_users)

    intersection = food_set.intersection(dessert_set)
    total_unique_users = food_set.union(dessert_set)

    probability = len(intersection) / len(total_unique_users)

    return probability



food_users = ['A', 'B', 'C', 'D', 'E']
dessert_users = ['B', 'C', 'E', 'F']

probability = calculate_probability(food_users, dessert_users)

print("Probability that a randomly selected user ordered both food and dessert:", probability)
```
