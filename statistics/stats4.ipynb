```python
import statistics

followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]

mean = statistics.mean(followers)
median = statistics.median(followers)

print("Mean:", mean)
print("Median:", median)
#the data is right-skewed (positively skewed).
```
