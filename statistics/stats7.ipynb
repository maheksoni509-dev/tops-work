```python
import matplotlib.pyplot as plt

instagram_hours = [2, 3, 4, 5, 4, 6, 7]

posts = [1, 1, 2, 2, 3, 3,4]
plt.scatter(instagram_hours, posts)

plt.xlabel("Instagram Hours per Day")
plt.ylabel("Number of Posts per Day")
plt.title("Instagram Usage vs Number of Posts")

plt.show()

# Observation:
# The scatter plot shows a positive correlation.
# As Instagram usage hours increase, the number of posts also increases.
```

```python
import numpy as np

zomato_orders = [10, 15, 20, 25, 30, 35]

swiggy_orders = [8, 12, 18, 22, 27, 32]

correlation = np.corrcoef(zomato_orders, swiggy_orders)[0, 1]

print("Correlation coefficient:", correlation)

# The correlation coefficient is close to +1.
# This means that friends who order more from Zomato
# generally also order more from Swiggy.
```
