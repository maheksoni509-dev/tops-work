```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

categories = ["A", "B", "C", "D"]
values = [25, 40, 15, 20]

scatter_x = [5, 10, 15, 20, 25]
scatter_y = [8, 18, 12, 22, 28]

pie_labels = ["Python", "Java", "C++", "JavaScript"]
pie_sizes = [35, 25, 20, 20]

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

ax[0, 0].plot(x, y, marker='o')
ax[0, 0].set_title("Line Chart")

ax[0, 1].bar(categories, values)
ax[0, 1].set_title("Bar Chart")

ax[1, 0].scatter(scatter_x, scatter_y)
ax[1, 0].set_title("Scatter Plot")

ax[1, 1].pie(pie_sizes, labels=pie_labels, autopct="%1.1f%%")
ax[1, 1].set_title("Pie Chart")

plt.tight_layout()

plt.show()
```

```python

apps = ["Zomato", "Swiggy", "Domino's"]
delivery_time = [30, 25, 35]   # Average delivery time in minutes

plt.figure(figsize=(6,5))

bars = plt.bar(
    apps,
    delivery_time,
    color=["red", "orange", "blue"],
    edgecolor="black",
    linewidth=2
)

bars[0].set_linestyle("-")    
bars[1].set_linestyle("--")   
bars[2].set_linestyle(":")    

plt.title("Average Delivery Time Comparison")
plt.xlabel("Food Delivery Apps")
plt.ylabel("Delivery Time (Minutes)")

plt.show()
```

```python

influencers = ["A", "B", "C", "D", "E"]
followers = [120, 250, 180, 300, 220]   
daily_posts = [2, 4, 3, 5, 2]           

fig, ax1 = plt.subplots(figsize=(8, 5))

ax1.plot(influencers, followers,
         color="blue", marker="o", linewidth=2,
         label="Followers")
ax1.set_xlabel("Influencers")
ax1.set_ylabel("Followers (Thousands)", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")

ax2 = ax1.twinx()
ax2.plot(influencers, daily_posts,
         color="red", marker="s", linewidth=2,
         label="Daily Posts")
ax2.set_ylabel("Average Daily Posts", color="red")
ax2.tick_params(axis="y", labelcolor="red")

plt.title("Instagram Followers vs Average Daily Posts")

plt.show()
```

```python

movies = ["Jawan", "Animal", "Stree 2", "Sikandar", "Bhool Bhulaiyaa 3"]
tickets_sold = [120, 95, 110, 85, 100]   # In lakhs
imdb_ratings = [7.0, 6.2, 8.0, 5.5, 6.8]

plt.figure(figsize=(8, 5))
plt.scatter(tickets_sold, imdb_ratings, color="purple", marker="o", s=100)

for i in range(len(movies)):
    plt.annotate(
        movies[i],
        (tickets_sold[i], imdb_ratings[i]),
        xytext=(0, 8),
        textcoords="offset points",
        ha="center"
    )

plt.title("Tickets Sold vs IMDB Ratings of Bollywood Movies")
plt.xlabel("Tickets Sold (Lakhs)")
plt.ylabel("IMDB Rating")
plt.grid(True)

plt.show()
```

```python

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 180, 170, 210, 190]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(months, sales, marker="o", color="blue")

max_sales = max(sales)
max_index = sales.index(max_sales)

ax.annotate(
    "Big Billion Days",
    (months[max_index], max_sales),
    xytext=(0, 15),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->", color="red")
)

ax.set_title("Flipkart Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales (in Crores)")

plt.show()
```
