```python

import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

delivery_time = np.random.randint(15, 61, 50)   # Random values between 15 and 60 minutes

sns.histplot(delivery_time, bins=10, kde=True)

plt.title("Distribution of Zomato Delivery Times")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.boxplot(x="day", y="total_bill", data=tips)

plt.title("Distribution of Total Bill by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill")

plt.show()
```

```python
import pandas as pd

teams = ["CSK", "MI", "RCB", "KKR", "SRH", "RR", "DC", "PBKS"]

data = {
    "Team": np.repeat(teams, 20),
    "Runs": np.random.randint(120, 231, 160)
}

df = pd.DataFrame(data)

sns.violinplot(x="Team", y="Runs", data=df)

plt.title("IPL Team Run Distribution")
plt.xlabel("Teams")
plt.ylabel("Runs")

plt.show()
```

```python

import random

genres = ["Pop", "Rock", "Hip-Hop", "Classical"] * 10
random.shuffle(genres)

df = pd.DataFrame({"Genre": genres})

sns.countplot(x="Genre", data=df)

plt.title("Number of Spotify Songs by Genre")
plt.xlabel("Genre")
plt.ylabel("Number of Songs")

plt.show()
```

```python

steps = [4500, 6200, 7000, 8500, 9000, 10500, 7800]

sns.set_theme(style="whitegrid")

sns.kdeplot(steps, fill=True, color="green")

plt.title("Distribution of Daily Step Counts")
plt.xlabel("Steps")
plt.ylabel("Density")

plt.show()
```
