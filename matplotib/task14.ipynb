```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.pairplot(tips)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

flights = sns.load_dataset("flights")

flight_data = flights.pivot_table(
    index="month",
    columns="year",
    values="passengers"
)

sns.heatmap(flight_data, annot=True, cmap="YlOrRd")

plt.title("Flight Passengers Heatmap")

plt.show()
```

```python

fmri = sns.load_dataset("fmri")

sns.relplot(
    data=fmri,
    x="timepoint",
    y="signal",
    hue="event",
    kind="line"
)

plt.title("FMRI Signal Over Time for Different Event Types")

plt.show()
```

```python

titanic = sns.load_dataset("titanic")

sns.catplot(
    data=titanic,
    x="sex",
    y="survived",
    kind="bar",
    ci=95
)

plt.title("Survival Rate by Gender")

plt.show()
```

```python

penguins = sns.load_dataset("penguins")

sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="flipper_length_mm",
    kind="reg"
)

plt.show()
```
