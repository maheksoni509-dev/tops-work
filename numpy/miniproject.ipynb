```python
import numpy as np

prices=np.array([100,200,150,300,np.nan,250])

quantity=np.array([10,5,8,3,7,6])

print("prices", prices)
print("quantity", quantity)

```

```python
print("missing values in prices",np.isnan(prices))

avg_price=np.nanmean(prices)
print("avg price",avg_price)
prices=np.where(np.isnan(prices),avg_price,prices)
print("cleaned prices", prices)
```

```python
revenue=prices*quantity
print("revenue for each product",revenue)
```

```python
print("total revenue",np.sum(revenue))
print("average revenue",np.mean(revenue))
print("maximum revenue",np.max(revenue))
print("minimumrevenue",np.min(revenue))
```

```python
bestproduct_index=np.argmax(revenue)
print("best selling product:",bestproduct_index)
print("revenue generated",revenue[bestproduct_index])
```

```python
sorted_indices=np.argsort(revenue)
print("sorted product indices",sorted_indices)
print("revenue sorted",revenue[sorted_indices])
```

```python
high_revenue_products=np.where(revenue>1000)
print("indices with revenue >1000",high_revenue_products
)
print("revenue values",revenue[high_revenue_products])
```

```python
print("number of products",prices.size)
print("total revenue",np.sum(revenue))
print("avg price",np.mean(prices))
print("avg quantity sold",np.mean(quantity))
```
