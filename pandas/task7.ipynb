```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\mahek\Downloads\zomato\zomato.csv",
    encoding="latin1"
)

print("Before Filling:")
print(df["Aggregate rating"].head(15))

df["Aggregate rating"] = df["Aggregate rating"].ffill()

df["Aggregate rating"] = df["Aggregate rating"].bfill()


print("\nAfter Filling:")
print(df["Aggregate rating"].head(15))
```
