```python
import pandas as pd
from pathlib import Path

data_folder = Path(".")

orders_file = data_folder / "orders.csv"
users_file = data_folder / "users.csv"

orders_df = pd.read_csv(orders_file)
users_df = pd.read_csv(users_file)

merged_df = pd.merge(users_df, orders_df, on="user_id")


result = merged_df[["username", "amount"]]

print(result)
```

```python

today_orders = pd.DataFrame({
    "order_id": [101, 102, 103],
    "item": ["Pizza", "Burger", "Pasta"],
    "price": [250, 180, 300]
})
yesterday_orders = pd.DataFrame({
    "order_id": [98, 99, 100],
    "item": ["Sandwich", "Fries", "Cold Coffee"],
    "price": [120, 90, 150]
})
combined_df = pd.concat([today_orders, yesterday_orders], ignore_index=True)


print(combined_df)
```
