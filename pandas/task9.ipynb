```python

df = pd.DataFrame({
    "username": [
        "insta_queen",
        "insta-queen",
        "instaqueen",
        "cool_user",
        "insta_queen",
        "travel_girl"
    ]
})

print("Original DataFrame:")
print(df)

df["username"] = df["username"].replace({
    "insta_queen": "instaqueen",
    "insta-queen": "instaqueen"
})

print("\nStandardized DataFrame:")
print(df)
```

```python

df = pd.DataFrame({
    "payment_status": [
        "Yes", "yes", "Y", "No", "no", "N",
        " Yes ", " no ", " y ", " n "
    ]
})

print("Original DataFrame:")
print(df)

df["payment_status"] = (
    df["payment_status"]
      .str.strip()      
      .str.lower()     
      .map({
          "yes": 1,
          "y": 1,
          "no": 0,
          "n": 0
      })
)

print("\nCleaned DataFrame:")
print(df)
```
