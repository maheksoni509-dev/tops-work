```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [6500, 7200, 8100, 5900, 9000, 10500, 7800]

plt.figure(figsize=(8, 5))
plt.plot(days, steps, marker='o')

plt.title("Daily Steps Over Last 7 Days")
plt.xlabel("Days")
plt.ylabel("Number of Steps")

plt.savefig("steps_lineplot.png")

plt.show()
```

```python
import matplotlib.pyplot as plt

restaurants = [
    "Cafe A", "Spice Hub", "Food Point", "BBQ Town", "Pizza House",
    "Royal Dine", "Burger Hub", "Green Bowl", "Tandoori Treat", "Urban Cafe"
]

ratings = [4.2, 3.8, 4.5, 4.0, 3.9, 4.7, 4.1, 3.7, 4.3, 4.6]
meal_price = [300, 250, 450, 350, 280, 600, 320, 220, 400, 550]

plt.figure(figsize=(8,5))
plt.scatter(ratings, meal_price)

plt.title("Zomato Ratings vs Average Meal Price")
plt.xlabel("Zomato Rating")
plt.ylabel("Average Meal Price (₹)")

plt.show()
```

```python

apps = ["Swiggy", "Zomato", "Domino's"]
orders = [12, 18, 7]

plt.figure(figsize=(7,5))
plt.bar(apps, orders, color=["orange", "red", "blue"], label="Orders")

plt.title("Orders Placed Last Month")
plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")

plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

durations = [15, 20, 35, 40, 18, 25, 30, 45, 50, 22,
             28, 32, 38, 41, 55, 60, 26, 34, 48, 52]

plt.figure(figsize=(8,5))
plt.hist(durations, bins=5, color="skyblue")

plt.title("Spotify Listening Session Durations")
plt.xlabel("Duration (minutes)")
plt.ylabel("Frequency")

plt.show()
```

```python

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

screen_time = [2.5, 3.0, 2.8, 3.5, 4.0, 5.0, 4.2]   
likes = [20, 35, 28, 40, 50, 60, 45]               

fig, ax = plt.subplots(1, 2, figsize=(12,5))

ax[0].plot(days, screen_time, marker="o", color="blue")
ax[0].set_title("Instagram Screen Time")
ax[0].set_xlabel("Days")
ax[0].set_ylabel("Hours")

ax[1].bar(days, likes, color="green")
ax[1].set_title("Posts Liked")
ax[1].set_xlabel("Days")
ax[1].set_ylabel("Number of Likes")

plt.savefig("social_media_usage.png")

plt.show()
```
