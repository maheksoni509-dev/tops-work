```python
import pandas as pd

df = pd.read_csv("zomato.csv", encoding="latin1")
print(df.columns)
```

```python
print(df["City"].unique())
```

```python
mumbai = df[df["City"] == "Mumbai"]
delhi = df[df["City"] == "New Delhi"]

mumbai.to_csv("mumbai_zomato.csv", index=False)
delhi.to_csv("delhi_zomato.csv", index=False)
```

```python
import sweetviz as sv

report = sv.compare([mumbai, "Mumbai"], [delhi, "Delhi"])
report.show_html("zomato_comparison_report.html")
```
