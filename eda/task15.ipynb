```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\mahek\Downloads\aaa\flipkart_com-ecommerce_sample.csv")


rating_counts = df['product_rating'].value_counts().sort_index()

plt.figure(figsize=(6,4))
sns.barplot(x=rating_counts.index, y=rating_counts.values)

plt.title("Count of Reviews by Rating")
plt.xlabel("Rating (Stars)")
plt.ylabel("Number of Reviews")

plt.show()
```

```python

df = pd.read_csv(r"C:\Users\mahek\Downloads\mydata\zomato.csv")

df['rate'] = df['rate'].str.replace('/5', '', regex=False)
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')

df['approx_cost(for two people)'] = (
    df['approx_cost(for two people)']
    .astype(str)
    .str.replace(',', '', regex=False)
)

df['approx_cost(for two people)'] = pd.to_numeric(
    df['approx_cost(for two people)'],
    errors='coerce'
)

df = df.dropna(subset=['approx_cost(for two people)', 'rate'])

plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x='approx_cost(for two people)',
    y='rate'
)

plt.title("Relationship Between Average Cost for Two and User Rating")
plt.xlabel("Average Cost for Two (₹)")
plt.ylabel("User Rating")
plt.show()
```

```python
import pandas as pd

songs_data = pd.read_json("song.json")

avg_streams = (
    songs_data.groupby('Genre')['Streams']
              .mean()
              .sort_values(ascending=False)
              .reset_index()
)

avg_streams.columns = ['Genre', 'Average Streams']

print(avg_streams)
```
